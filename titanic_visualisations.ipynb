# Titanic visualisations

For each variable, the first graph compares train vs test data and the second shows survival.

## Setup

```python
from titanic.data import combine_train_test, load_data
from titanic.features import add_age_interval, add_fare_interval
from titanic.plots import plot_counts, plot_distribution, save_figure
```

```python
# Load data, combine train and test, and add the interval features
train_df, test_df = load_data()
all_df = combine_train_test(train_df, test_df)
all_df = add_age_interval(all_df)
all_df = add_fare_interval(all_df)
```

## Sex

```python
# Number of passengers per Sex: train vs test
with save_figure("01_sex_train_vs_test.png"):
    plot_counts(all_df, "Sex", hue="set")
```

```python
# Number of passengers per Sex: survived vs died
with save_figure("02_sex_survival.png"):
    plot_counts(all_df, "Sex", hue="Survived")
```

## Pclass

```python
# Number of passengers per Pclass: train vs test
with save_figure("03_pclass_train_vs_test.png"):
    plot_counts(all_df, "Pclass", hue="set")
```

```python
# Plot count pairs using all_df for Pclasss
# (Test rows have no Survived value, so they are left out of these plots)
plot_counts(all_df, "Pclass", hue="Survived")
```

## Age interval

```python
# Distribution of Age: train vs test
plot_distribution(all_df, "Age", hue="set")
```

```python
# Distribution of Age: survival
plot_distribution(all_df, "Age", hue="Survived")
```

```python
# Number of passengers per Age Interval: survived vs died
with save_figure("04_age_interval_survival.png"):
    plot_counts(all_df, "Age Interval", hue="Survived")
```

## Fare

```python
# Distribution of Fare: train vs test
with save_figure("05_fare_train_vs_test.png"):
    plot_distribution(all_df, "Fare", hue="set")
```

```python
# Number of passengers per Fare Interval: survived vs died
with save_figure("06_fare_interval_survival.png"):
    plot_counts(all_df, "Fare Interval", hue="Survived")
```

## Multivariate: Age interval and Pclass

```python
# Number of passengers per Age Interval, split by Pclass
with save_figure("07_age_interval_by_pclass.png"):
    plot_counts(all_df, "Age Interval", hue="Pclass")
```
