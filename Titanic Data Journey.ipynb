## Process names


When we process names, we would like to extract the following information:

- Family name - this is the first word (or few first words, if a family name with multiple names), followed by a comma  
- Title - this follows just after the comma   
- Given name - this is the word or group of words following family name  
- Maiden name - for ladies, is given between parantheses  

We start with creating a function that parses the Name string and extract (if possible) these 4 elements



```python
# Split Name into Family Name, Title, Given Name and Maiden Name
all_df = add_name_features(all_df)
train_df = add_name_features(train_df)
```

### Few more engineered data 


Let's create two more engineered features:  
* Family size interval: Single, Small, Large  
* Aggregated titles: Mr, Mrs, Master, Miss, and Rare  

```python
# Label each family as Single (1 person), Small (2-4) or Large (5 or more)
all_df = add_family_type(all_df)
train_df = add_family_type(train_df)
```

```python
# Tidy the titles: merge Mlle/Ms/Mme into Miss/Mrs and group uncommon titles as 'Rare'
all_df = add_aggregated_titles(all_df)
train_df = add_aggregated_titles(train_df)
```

```python
# Average survival rate for each combination of title and sex (training data)
survival_rate(train_df, ["Titles", "Sex"])
```

# Baseline model

## Feature engineering: Map categorical value to numerical values

```python
# Replace 'female'/'male' with 1/0 for the model (done after Sex_Pclass, which needs the text)
train_df = encode_sex(train_df)
test_df = encode_sex(test_df)
```

Create the train-validation split.

```python
# Split the training data into training and validation sets (X = Sex and Pclass, Y = Survived)
train_X, train_Y, valid_X, valid_Y = split_data(train_df)
```

Train the random forest classifier on the training data.

```python
# Train a random forest classifier on the training set
clf = train_model(train_X, train_Y)
```

# Evaluation

## Classification report for training data.

```python
# Classification report on the training data (shows the training error)
evaluate(clf, train_X, train_Y)
```

Classification report for validation data.

```python
# Classification report on the validation data (shows how well the model does on unseen rows)
evaluate(clf, valid_X, valid_Y)
```
